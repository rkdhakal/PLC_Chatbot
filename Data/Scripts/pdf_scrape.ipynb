```python
import pdfplumber
import pandas as pd

# Define the PDF file path
pdf_path = "../AB 830 micrologix.pdf"

# Initialize a list to store tables
tables = []

# Open the PDF file using pdfplumber
with pdfplumber.open(pdf_path) as pdf:
    for page_num in range(309, 314):  # Iterate through the specified pages
        page = pdf.pages[page_num - 1]  # Access the page (pdfplumber is 0-indexed)

        # Extract table using optimized parameters
        extracted_table = page.extract_table({
            "horizontal_strategy": "lines",  # Better table detection
            "vertical_strategy": "lines",  # Ensure proper column separation
            "snap_tolerance": 3,  # Helps align table structure
            "intersection_x_tolerance": 5  # Ensures proper column breaks
        })

        # Debug output: Print the extracted table
        if extracted_table:
            print(f"Page {page_num} Extracted Table:")
            for row in extracted_table:
                print(row)  # Print each row to debug structure
            tables.extend(extracted_table)  # Append extracted data

# Convert extracted table into a Pandas DataFrame
if tables:
    df = pd.DataFrame(tables[1:], columns=tables[0])  # First row is headers
    df.to_csv("extracted_table.csv", index=False)  # Save table to CSV
    print(df)
else:
    print("No tables extracted.")
```
